```python
import pandas as pd
```

```python
import csv
```

```python
import os
```

```python
data_to_write = [
    ["ID", "Name", "Age", "Grade"],
    [1,"Alice", 20, "A"],
    [2,"Bob", 22, "B"],
    [3,"Charlie", 21, "C"],
    [4,"Diana",23,"A"]
]
```

```python
file_name = "students.csv"
```

```python
with open(file_name, mode='w',newline='',encoding='utf-8') as file:
  writer = csv.writer(file)
  writer.writerows(data_to_write)
```

```python
print(f"Successfully created and wrote data to '{file_name}'!\n")
```

```python
print("--- Method A: Loading into Python List of Dictionaries ---")
python_data_structure = []

with open(file_name, mode='r', encoding='utf-8') as file:
    csv_reader = csv.DictReader(file)
    for row in csv_reader:
        python_data_structure.append(row)

print(python_data_structure)
print(f"Loaded {len(python_data_structure)} rows into a Python list.\n")


# Method B: Loading into a Pandas DataFrame (The industry-standard structure for data analysis)
print("--- Method B: Loading into a Pandas DataFrame ---")
df = pd.read_csv(file_name)

# Displaying the internal DataFrame structure
print(df)
print("\nData types of columns:")
print(df.dtypes)
```
