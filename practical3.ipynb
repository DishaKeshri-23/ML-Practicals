```python
import pandas as pd
import numpy as np

# 1. Create a sample dirty dataset for demonstration
raw_data = {
    "User_ID": [101, 102, 103, 103, 104, 105, 106],
    "Full_Name": ["  Alice Johnson ", "Bob Smith", "Charlie Brown", "Charlie Brown", "  diana Prince", "Evan Wright ", "Frank White"],
    "Age": ["25", "30", np.nan, np.nan, "22", "unknown", "200"],  # Contains nulls, string types, and outlier
    "Salary": [50000.0, 60000.0, 75000.0, 75000.0, np.nan, 80000.0, -5000.0],  # Contains negative value and null
    "Join_Date": ["2021-01-15", "2020/05/20", "2019-11-02", "2019-11-02", "15-08-2022", "InvalidDate", "2023-03-10"],
    "Department": ["IT", "hr", "Finance", "Finance", "it", "HR", "Sales"]
}

df = pd.DataFrame(raw_data)
print("=== ORIGINAL DIRTY DATA ===")
print(df)
print("\n" + "=" * 50 + "\n")

# ---------------------------------------------------------
# STEP 1: Handle Duplicates
# ---------------------------------------------------------
# Remove exact duplicate rows
df = df.drop_duplicates()

# ---------------------------------------------------------
# STEP 2: Standardize and Clean Text Columns
# ---------------------------------------------------------
# Strip whitespace and normalize casing
df["Full_Name"] = df["Full_Name"].str.strip().str.title()
df["Department"] = df["Department"].str.strip().str.upper()

# ---------------------------------------------------------
# STEP 3: Handle Invalid Data Types & Missing Values
# ---------------------------------------------------------
# Convert 'Age' to numeric (invalid strings like 'unknown' become NaN)
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

# Impute missing Age with median value
df["Age"] = df["Age"].fillna(df["Age"].median())

# Impute missing Salary with mean value
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

# Parse dates (unparseable values become NaT)
df["Join_Date"] = pd.to_datetime(df["Join_Date"], errors="coerce")
# Fill missing dates with a placeholder or drop them
df["Join_Date"] = df["Join_Date"].fillna(pd.to_datetime("2020-01-01"))

# ---------------------------------------------------------
# STEP 4: Handle Outliers and Invalid Values
# ---------------------------------------------------------
# Cap Age between realistic bounds (18 to 100)
df.loc[df["Age"] > 100, "Age"] = df["Age"].median()
df.loc[df["Age"] < 18, "Age"] = 18

# Replace negative salaries with the column median
df.loc[df["Salary"] < 0, "Salary"] = df["Salary"].median()

# Cast columns to their final correct data types
df["Age"] = df["Age"].astype(int)
df["Salary"] = df["Salary"].round(2)

# Reset index after dropping rows
df = df.reset_index(drop=True)

# ---------------------------------------------------------
# STEP 5: Final Output & Verification
# ---------------------------------------------------------
print("=== CLEANED DATA ===")
print(df)
print("\n=== DATA TYPES & INFO ===")
print(df.info())
```
